# ML Experiment - 1
# Simple Linear Regression from Scratch


```python
# Step 1: Import required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Step 2: Create Dataset
data = {
    'X': [2, 3, 4, 5, 6, 7, 8, 9, 10],
    'Y': [1, 3, 6, 9, 11, 13, 15, 17, 20]
}

df = pd.DataFrame(data)
```

```python
# Step 3: Extract independent and dependent variables
X = df['X']
Y = df['Y']

# Number of data points
n = len(df)
```

```python
# Step 4: Calculate mean of X and Y
x_mean = np.mean(X)
y_mean = np.mean(Y)
```

```python
# Step 5: Calculate slope (Beta_1)
numerator_beta1 = np.sum((X - x_mean) * (Y - y_mean))
denominator_beta1 = np.sum((X - x_mean) ** 2)
beta1 = numerator_beta1 / denominator_beta1

# Step 6: Calculate intercept (Beta_0)
beta0 = y_mean - beta1 * x_mean

# Display regression coefficients
print("Beta_0 (Intercept):", beta0)
print("Beta_1 (Slope):", beta1)
```

```python
# Step 7: Scatter plot of actual data
plt.figure(figsize=(10, 6))
plt.scatter(X, Y, color='blue')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Scatter Plot of X vs Y')
plt.show()
```

```python
# Step 8: Predict Y values
y_predicted = beta0 + beta1 * X

# Step 9: Plot regression line
plt.figure(figsize=(10, 6))
plt.scatter(X, Y, color='blue', label='Actual Data')
plt.plot(X, y_predicted, color='red', label='Regression Line')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Linear Regression Model')
plt.legend()
plt.show()
```

```python
# Step 10: Calculate error metrics
MSE = np.sum((Y - y_predicted) ** 2) / n
RMSE = np.sqrt(MSE)

# Display errors
print("Mean Squared Error (MSE):", MSE)
print("Root Mean Squared Error (RMSE):", RMSE)
```

```python
# Step 11: Display predicted values
print("\nPredicted Y values:")
print(y_predicted)
```
